```python
from pathlib import Path
guard = PipelineGuard(
    policy=ErrorPolicy.LOG,
    dedup=True,
    checkpoint_path=Path("corpus.ckpt.jsonl"),
    checkpoint_every=200,
)
for doc in guard.iter(reader.get_documents()):
    process(doc)
guard.close()
```
