Regular event data:

Measured point data with errors:


````python
t = 100 * np.random.random(100)
x = np.exp(-0.5 * (t - 50) ** 2)
sigma = 0.1
x_obs = np.random.normal(x, sigma)
edges = bayesian_blocks(t, x_obs, sigma, fitness='measures')
````
