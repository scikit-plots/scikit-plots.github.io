````python
from scikitplot.corpus import evaluate_retrieval
def retriever(query):
    return [("d1", 0.9), ("d2", 0.4)] if "cat" in query else [("d3", 0.7)]
report = evaluate_retrieval(
    retriever,
    queries={"q1": "cat food", "q2": "dog toys"},
    qrels={"q1": ["d2"], "q2": ["d4"]},
    metrics=["recall@2", "mrr"],
)
report.metrics
````
