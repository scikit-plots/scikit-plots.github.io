def my_para_chunk(text, metadata):
paras = [p.strip() for p in text.split("\\n\\n") if p.strip()]
cursor = 0
result = []
for para in paras:
idx = text.find(para, cursor)
result.append((idx, para))
cursor = idx + len(para)
return result
