```python
import random
from scikitplot.cexternals._annoy import Annoy, AnnoyIndex

f=100
n=1000
idx = AnnoyIndex(f, metric='l2')

for i in range(n):
   v = [random.gauss(0, 1) for _ in range(f)]
   idx.add_item(i, v)
```
